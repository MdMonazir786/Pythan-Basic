3. What are pre-defined keywords in Python?

12. How does the elif statement work?
-> elif stands for “else if”.

How it works:

a. Python checks the if condition first.

b. If if is False, it checks the elif condition.

13. What is the difference between for and while loops?
-> Loop : A loop is used to repeat a block of code multiple times — until a condition is met or a sequence ends.

“for” loops are used for iterating over a sequence of elements, while “while” loops are used for executing code until a specified condition becomes False.

14.  Describe a scenario where a while loop is more suitable than a for loop?
-> Scenario: Keep asking for correct password until it's right.

Imagine you're building a login system. You want to keep asking the user to enter the correct password, and you don't know how many tries they’ll need.

While loop: Because we don’t know how many times the user will enter the wrong password — so we loop until the condition is met.

For loop: With a for loop, you’d have to set a fixed number of attempts.

But in this scenario, we want to allow unlimited attempts — so while is more flexible.

Conclusion: Use a while loop when you don’t know in advance how many times something should repeat — just that it should keep going while a condition is true.

```python
# Create a list and tuple in Python, and demonstrate how attempting to change an element works differently for each.

# Create a list
my_list = [10, 20, 30]
print("Original list:", my_list)

# Modify an element in the list
my_list[1] = 99
print("Modified list:", my_list)


# Create a tuple
my_tuple = (10, 20, 30)
print("\nOriginal tuple:", my_tuple)

# Try to modify an element in the tuple
try:
    my_tuple[1] = 99
except TypeError as e:
    print("Error when trying to modify tuple:", e)


```

```python
# Write a function to demonstrate the behavior of mutable and immutable arguments.

def modify_arguments(num, num_list):
    print("\nInside function:")

    # Try to modify the integer (immutable)
    num += 10
    print("Modified num (int):", num)

    # Modify the list (mutable)
    num_list.append(100)
    print("Modified list:", num_list)


# Outside the function
x = 5
my_list = [1, 2, 3]

print("Before function call:")
print("x =", x)
print("my_list =", my_list)

# Call the function
modify_arguments(x, my_list)

print("\nAfter function call:")
print("x =", x)              # Stays the same
print("my_list =", my_list)  # Gets modified
```

```python
# Write a program that performs basic arithmetic operations on two user-input numbers.

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Perform operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2

# Handle division safely
if num2 != 0:
    division = num1 / num2
else:
    division = "Undefined (cannot divide by zero)"

# Display results
print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)

```

```python
#  Write a program to demonstrate the use of logical operators.

a = int(input("Enter first number (a): "))
b = int(input("Enter second number (b): "))

print("\nLogical Operations:")

# AND operator
if a > 0 and b > 0:
    print("Both a and b are positive numbers.")

# OR operator
if a > 0 or b > 0:
    print("At least one number is positive.")

# NOT operator
if not (a < 0 or b < 0):
    print("Neither a nor b is negative.")
else:
    print("At least one number is negative.")
```

```python
#  Write a Python program to convert user input from string to integer, float, and boolean types.

# input as a string
user_input = input("Enter a value: ")

# Convert to integer
try:
    int_value = int(user_input)
    print("Integer:", int_value)
except ValueError:
    print("Cannot convert to Integer ❌")

# Convert to float
try:
    float_value = float(user_input)
    print("Float:", float_value)
except ValueError:
    print("Cannot convert to Float ❌")

# Convert to boolean
bool_value = bool(user_input)
print("Boolean:", bool_value)
```

```python
# Write code to demonstrate type casting with list elements.

# List of string numbers
string_list = ['10', '20', '30', '0', '']

print("Original list:", string_list)

# Convert to integers
try:
    int_list = [int(item) for item in string_list if item != '']
    print("List as integers:", int_list)
except ValueError as e:
    print("Error converting to integers:", e)

# Convert to floats
try:
    float_list = [float(item) for item in string_list if item != '']
    print("List as floats:", float_list)
except ValueError as e:
    print("Error converting to floats:", e)

# Convert to booleans
bool_list = [bool(item) for item in string_list]
print("List as booleans:", bool_list)
```

```python
 # Write a Python program to find the sum of all even numbers between 1 and 50.

 # Initialize the sum
even_sum = 0

# Loop through numbers from 1 to 50
for num in range(1, 51):
    if num % 2 == 0:
        even_sum += num

# Print the result
print("The sum of even numbers from 1 to 50 is:", even_sum)
```

```python
#  Write a program to reverse a string using a while loop.

text = input("Enter a string: ")

# Initialize index and reversed string
index = len(text) - 1
reversed_text = ""

# Use while loop to build the reversed string
while index >= 0:
    reversed_text += text[index]
    index -= 1

# Print the reversed string
print("Reversed string:", reversed_text)
```

```python
# Write a Python program to calculate the factorial of a number provided by the user using a while loop.

num = int(input("Enter a non-negative integer: "))

# Check for negative input
if num < 0:
    print("Factorial is not defined for negative numbers.")
else:
    factorial = 1
    i = 1

    while i <= num:
        factorial *= i
        i += 1

    print(f"The factorial of {num} is: {factorial}")
```
